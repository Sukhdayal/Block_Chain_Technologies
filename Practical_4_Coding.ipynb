# Practical 4: Program to Solve a Mining Puzzle Using Blockchain



## Step 1: Import the Required Library

````python
# Import hashlib for generating SHA-256 hash
import hashlib
````

### What this step does

We import `hashlib` because we need SHA-256 to generate the hash used in our mining puzzle.

## Step 2: Define the Mining Puzzle Data

````python
# Define the data for the mining puzzle
data = "Rahul sends 2 coins to Priya"

print("Data:", data)
````

### What this step does

This is the data that will be combined with different nonce values and passed to SHA-256.

## Step 3: Set the Mining Difficulty

````python
# Difficulty 4 means the hash must start with four zeros
difficulty = 4

# Create the target according to the difficulty
# Example: difficulty = 4 → target = "0000"
target = "0" * difficulty

print("Difficulty:", difficulty)
print("Target:", target)
````

### What this step does

A difficulty of `4` means that a valid hash must begin with `0000`. The `target` variable stores this required pattern.

## Step 4: Initialize the Nonce

````python
# Start searching from nonce 0
nonce = 0

print("Starting nonce:", nonce)
````

### What this step does

The nonce is the number that the miner will keep changing. We begin the search from `0`.

## Step 5: Display Mining Information

````python
print("========== MINING PUZZLE ==========")
print("Data:", data)
print("Difficulty:", difficulty)
print("Target:", target)
print()
````

### What this step does

This step simply displays the information about our mining puzzle before mining starts.

## Step 6: Start the Mining Loop

````python
# Keep trying until a valid hash is found
while True:
    print("Trying nonce:", nonce)

    # Stop here for now while learning the structure.
    # The actual mining operations are added in the next cells.
    break
````

### What this step does

`while True` creates a loop that can continue indefinitely. In the complete program, the loop will stop only when a valid hash is found using `break`.

The temporary `break` in this teaching cell is only to let us see the loop structure without starting the full mining process yet.

## Step 7: Combine Data and Nonce

````python
# Example of combining the data with the current nonce
text = data + str(nonce)

print("Text used for hashing:")
print(text)
````

### What this step does

The current nonce is joined with the block data. This combined text becomes the input for SHA-256.

## Step 8: Generate the SHA-256 Hash

````python
# Generate SHA-256 hash
hash_value = hashlib.sha256(
    text.encode()
).hexdigest()

print("Hash:", hash_value)
````

### What this step does

The combined data and nonce are converted into a SHA-256 hash. We get a completely different hash when the nonce changes.

## Step 9: Check Whether the Hash Solves the Puzzle

````python
# Check whether the hash satisfies the target
if hash_value.startswith(target):
    print("Valid hash found!")
else:
    print("Hash is not valid yet.")
````

### What this step does

The program checks whether the hash starts with the required target, such as `0000`. If it does, the mining puzzle is solved. If not, we need another nonce.

## Step 10: Complete Mining Program

````python
# Start from nonce 0 again for the complete mining process
nonce = 0

# Keep trying different nonce values
while True:

    # Combine the data with the current nonce
    text = data + str(nonce)

    # Generate SHA-256 hash
    hash_value = hashlib.sha256(
        text.encode()
    ).hexdigest()

    # Check whether the hash satisfies the target
    if hash_value.startswith(target):

        # Valid hash found
        print("Mining Successful!")
        print("Nonce:", nonce)
        print("Hash:", hash_value)

        # Stop the loop
        break

    # If hash is not valid, try the next nonce
    nonce += 1
````

### Final Flow

```text
Data
  ↓
Nonce = 0
  ↓
Combine Data + Nonce
  ↓
SHA-256 Hash
  ↓
Does Hash start with 0000?
  ↓
 NO ──→ Increase Nonce ──→ Try Again
  ↓
 YES
  ↓
Mining Successful
```
