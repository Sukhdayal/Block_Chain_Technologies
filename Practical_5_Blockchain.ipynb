# Practical No. 5: Blockchain Transaction Processing and Analysis

## Aim
To formulate mathematical models and solve a real-time blockchain transaction processing problem using appropriate data structures, algorithms, and Python programming techniques.

### What we will build
A simple **Blockchain Transaction Processing System** that creates transactions, stores them, processes them using a Queue, calculates values and fees, searches transactions, finds the highest transaction, sorts transactions, creates a simplified block, and generates a SHA-256 hash.

## 1. Understanding the Problem

Consider a blockchain network receiving transactions. We will use five simple transactions:

| ID | Sender | Receiver | Amount |
|---|---|---|---:|
| T1 | Rahul | Priya | ₹500 |
| T2 | Aman | Rahul | ₹200 |
| T3 | Priya | Aman | ₹800 |
| T4 | Rahul | Aman | ₹300 |
| T5 | Aman | Priya | ₹600 |

The overall flow is:

**Transactions → Store → Process → Calculate → Search → Sort → Create Block → Hash**

## 2. Blockchain Transaction and Dictionary

A transaction records an action between participants. For example, **Rahul → Priya → ₹500**. We represent one transaction using a Python **dictionary**. A dictionary stores information as **key → value** pairs, such as `"from" → "Rahul"` and `"amount" → 500`.

```python
# Create one blockchain transaction using a dictionary.
transaction1 = {
    "id": "T1",          # Store the transaction ID.
    "from": "Rahul",     # Store the sender's name.
    "to": "Priya",       # Store the receiver's name.
    "amount": 500        # Store the transaction amount.
}

# Display the complete transaction.
print(transaction1)
```

## 3. Accessing Dictionary Values

Because the transaction is stored as a dictionary, we can access each part using its key. For example, `transaction1["amount"]` gives the amount.

```python
# Display individual values from the transaction dictionary.
print("Transaction ID:", transaction1["id"])       # Get the ID.
print("Sender:", transaction1["from"])             # Get the sender.
print("Receiver:", transaction1["to"])             # Get the receiver.
print("Amount: ₹", transaction1["amount"])         # Get the amount.
```

## 4. Data Structure: List

We have multiple transactions, so we store them in a **list**. A list lets us keep many transaction dictionaries together while preserving their order.

```python
# Create the remaining transactions.
transaction2 = {"id": "T2", "from": "Aman", "to": "Rahul", "amount": 200}  # T2
transaction3 = {"id": "T3", "from": "Priya", "to": "Aman", "amount": 800}  # T3
transaction4 = {"id": "T4", "from": "Rahul", "to": "Aman", "amount": 300}  # T4
transaction5 = {"id": "T5", "from": "Aman", "to": "Priya", "amount": 600}  # T5

# Store all transactions inside one list.
transactions = [
    transaction1,  # Add T1.
    transaction2,  # Add T2.
    transaction3,  # Add T3.
    transaction4,  # Add T4.
    transaction5   # Add T5.
]

# Display all transactions.
for transaction in transactions:
    print(transaction)
```

## 5. Data Structure: Queue

A **Queue** works like a line of people. The person who enters first is served first. This is called **FIFO: First In, First Out**.

For our transactions:

`T1 → T2 → T3 → T4 → T5`

We use Python’s `deque` to implement the Queue. `append()` adds an item at the end, while `popleft()` removes the first item.

```python
# Import deque to create a Queue.
from collections import deque

# Create an empty Queue.
transaction_queue = deque()

# Add every transaction to the Queue.
for transaction in transactions:
    transaction_queue.append(transaction)  # Add transaction at the end.

# Process transactions using FIFO.
while transaction_queue:
    transaction = transaction_queue.popleft()  # Remove the first transaction.
    print("Processing:", transaction["id"])    # Display the transaction.
```

## 6. Mathematical Model: Total Transaction Value

The total is the sum of all transaction amounts.

**Total = A₁ + A₂ + ... + Aₙ**

For our data: **500 + 200 + 800 + 300 + 600 = ₹2400**.

```python
# Start the total at zero.
total_amount = 0

# Go through every transaction.
for transaction in transactions:
    total_amount += transaction["amount"]  # Add the current amount.

# Display the total.
print("Total Transaction Value: ₹", total_amount)
```

## 7. Mathematical Model: Average Transaction Value

The average gives the typical transaction value.

**Average = Total / Number of Transactions**

For our data: **₹2400 / 5 = ₹480**.

```python
# Count the number of transactions.
number_of_transactions = len(transactions)  # len() gives the number of items.

# Calculate the average.
average_amount = total_amount / number_of_transactions

# Display the average.
print("Average Transaction Value: ₹", average_amount)
```

## 8. Mathematical Model: Transaction Fee

For this example, we use a **2% transaction fee**.

**Fee = Amount × (Fee Rate / 100)**

For ₹500: **₹500 × 2/100 = ₹10**.

```python
# Set the transaction fee rate to 2 percent.
fee_rate = 2

# Calculate the fee for every transaction.
for transaction in transactions:
    amount = transaction["amount"]                  # Get the amount.
    fee = amount * (fee_rate / 100)                  # Apply the fee formula.
    print(transaction["id"], "Amount: ₹", amount, "Fee: ₹", fee)  # Display result.
```

## 9. Algorithm: Linear Search

**Linear Search** checks items one by one until the required item or condition is found. It is like checking a classroom list from the first student onward.

Here we search for **all transactions involving Rahul**. The program checks T1, then T2, then T3, and so on. The matching transactions are **T1, T2, and T4**.

```python
# Select the user we want to search for.
search_user = "Rahul"

# Check every transaction one by one.
for transaction in transactions:
    # Check whether Rahul is the sender or receiver.
    if transaction["from"] == search_user or transaction["to"] == search_user:
        # Display the matching transaction.
        print(transaction["id"], ":", transaction["from"], "->", transaction["to"], ": ₹", transaction["amount"])
```

## 10. Algorithm: Finding the Maximum

To find the highest transaction, we first assume that the first transaction is the largest. Then we compare every other transaction with it. If we find a larger amount, we replace the current maximum.

For our data, the highest transaction is **T3 = ₹800**.

```python
# Assume the first transaction is the largest.
maximum_transaction = transactions[0]

# Check every transaction.
for transaction in transactions:
    # Compare the current amount with the maximum amount.
    if transaction["amount"] > maximum_transaction["amount"]:
        maximum_transaction = transaction  # Update the maximum.

# Display the result.
print("Highest Transaction:", maximum_transaction["id"])
print("Amount: ₹", maximum_transaction["amount"])
```

## 11. Algorithm: Bubble Sort

**Bubble Sort** repeatedly compares neighbouring elements. If they are in the wrong order, they are swapped.

Example: `500, 200` becomes `200, 500`.

For our transactions, we want the order:

`₹200 → ₹300 → ₹500 → ₹600 → ₹800`

Bubble Sort is used here because its internal working is easy for students to observe and understand.

```python
# Make a copy so that the original order is not changed.
sorted_transactions = transactions.copy()

# Find the number of transactions.
n = len(sorted_transactions)

# Start Bubble Sort.
for i in range(n):
    # Compare neighbouring transactions.
    for j in range(0, n - i - 1):
        # Check whether the current amount is greater than the next amount.
        if sorted_transactions[j]["amount"] > sorted_transactions[j + 1]["amount"]:
            # Swap the two transactions.
            sorted_transactions[j], sorted_transactions[j + 1] = sorted_transactions[j + 1], sorted_transactions[j]

# Display the sorted transactions.
for transaction in sorted_transactions:
    print(transaction["id"], ": ₹", transaction["amount"])
```

## 12. Creating a Simplified Blockchain Block

After processing and analyzing the transactions, we represent them as a simple blockchain block.

Our block contains:

- Block number
- Previous hash
- Transactions
- Total value
- Average value
- Highest transaction
- Current block hash

This is a **simplified block**, not a complete production blockchain.

```python
# Assign a block number.
block_number = 1

# For the first block, use a fixed previous hash value.
previous_hash = "0000000000000000"

# Display basic block information.
print("Block Number:", block_number)
print("Previous Hash:", previous_hash)

# Display the transactions inside the block.
for transaction in transactions:
    print(transaction["id"], ":", transaction["from"], "->", transaction["to"], ": ₹", transaction["amount"])
```

## 13. Generating the SHA-256 Block Hash

Hashing converts input data into a fixed-length hexadecimal value. We use **SHA-256** for our simplified block.

We combine the block information, convert it to bytes, and pass it to SHA-256. If the input changes, the resulting hash changes as well. This demonstrates the basic role of cryptographic hashing in blockchain.

```python
# Import hashlib for SHA-256.
import hashlib

# Start with an empty string for block data.
block_data = ""

# Add every transaction to the block data.
for transaction in transactions:
    block_data += transaction["id"] + transaction["from"] + transaction["to"] + str(transaction["amount"])

# Add the block number.
block_data += str(block_number)

# Add the previous hash.
block_data += previous_hash

# Generate the SHA-256 hash.
block_hash = hashlib.sha256(block_data.encode()).hexdigest()

# Display the hash.
print("Block Hash:", block_hash)
```
